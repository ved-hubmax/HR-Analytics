3. what is thr attrition rate of employees leaving the company having project more than 5

```python
more_project = df.loc[(df['numberOfProjects']> 5) & (df['left'] == 1)]
print(f" the number of people  {len(more_project)}")
print(f" percentage : {(len(more_project)/11991)*100}%")
```

```python
more_than_5 = df[df["numberOfProjects"] > 5]

# Compute attrition rate
attrition_rate = (more_than_5["left"] == 1).mean()

print("Employees with >5 projects:", len(more_than_5))
print("Employees who left:", (more_than_5["left"] == 1).sum())
print(f"Attrition rate: {attrition_rate:.2%}")
```

```python
df['project_group'] = pd.cut(df['numberOfProjects'],
                              bins=[0, 3, 5, df['numberOfProjects'].max()],
                              labels=['0-3 projects', '4-5 projects', '6+ projects'])


sns.countplot(data=df[df['left']==1], x='project_group' , color = 'green' , edgecolor = 'black')
plt.title('Employees Who Left by Number of Projects')
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.set_style('dark')
sns.countplot(
    x ='numberOfProjects',
    hue ='left',
    data = df,
    palette ='Set2',
    edgecolor ='black'
)

plt.title('Plot3 :  employees attrition rate of a company by number of project', fontsize = 14,fontweight = 'bold')
plt.xlabel('number of projects', fontsize = 12)
plt.ylabel('number of employee', fontsize = 14)
plt.legend(title ='attrition rate')
plt.xticks(rotation = 60)
plt.tight_layout()
plt.show()
```
